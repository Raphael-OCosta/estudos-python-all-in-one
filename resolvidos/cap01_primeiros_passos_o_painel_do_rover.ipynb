### 1.7 **Primeiro módulo do rover.** Escreva `painel.py` com as constantes `NOME = "Aurora"` e `VERSAO = "0.1"` e a função `status(bateria, temperatura, ligado)`, que devolve um texto de várias linhas: nome e versão, bateria com `%`, temperatura com 1 casa e `°C`, e o estado `LIGADO` / `DESLIGADO`.

_A célula grava `gabarito_rover/painel.py`: a 1ª linha (`%%writefile`) já vem pronta, digite o código embaixo dela._

```python
%%writefile ../gabarito_rover/painel.py
"""Painel de status do rover Aurora."""
NOME = "Aurora"
VERSAO = "0.1"


def status(bateria: int, temperatura: float, ligado: bool) -> str:
    estado = "LIGADO" if ligado else "DESLIGADO"
    return (f"Rover {NOME} v{VERSAO}\n"
            f"  bateria: {bateria}%\n"
            f"  temperatura: {temperatura:.1f} °C\n"
            f"  estado: {estado}")
```
